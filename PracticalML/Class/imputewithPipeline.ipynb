```python
from sklearn.linear_model import LinearRegression,ElasticNet,Ridge,Lasso
from sklearn.model_selection import train_test_split
from sklearn.compose import make_column_transformer, make_column_selector
from sklearn.metrics import r2_score,root_mean_squared_error
import pandas as pd
from sklearn.preprocessing import OneHotEncoder,StandardScaler
from sklearn.impute import SimpleImputer
```

```python
chemicals=pd.read_csv(r'E:\DataAnalytics\Cases\Chemical_Process_Data\ChemicalProcess.csv')
chemicals.head()
```

```python
chemicals.isnull().sum().sum()
```

```python
X=chemicals.drop('Yield',axis=1)
y=chemicals['Yield']
```

```python
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.3,random_state=26)
X_train.isnull().sum().sum()
```

```python
X_test.isnull().sum().sum()
```

```python
imputer=SimpleImputer(strategy='mean')
```

```python
model=LinearRegression()
```

```python
from sklearn.pipeline import Pipeline
pipe = Pipeline([('IMP',imputer),('LR',model)])
pipe.fit(X_train,y_train)
```

```python
y_pred=pipe.predict(X_test)
```

```python
print('R 2 score:->',r2_score(y_test,y_pred))
print('Root_Mean:->',root_mean_squared_error(y_test,y_pred))
```
