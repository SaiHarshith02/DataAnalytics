```python
df=pd.read_csv(r"E:\DataAnalytics\Cases\human-resources-analytics\HR_comma_sep.csv")
df.head()
```

```python
X,y=df.drop('left',axis=1),df['left']
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.3,random_state=26,stratify=y)
```

```python
from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import make_column_selector,make_column_transformer

encoder=OneHotEncoder(sparse_output=False, drop='first')
ct=make_column_transformer((encoder,make_column_selector(dtype_include='object')),
                          ('passthrough',make_column_selector(dtype_include='number')),
                           verbose_feature_names_out=False
                          )
```

```python
solvers = ['lbfgs', 'liblinear', 'saga', 'sag', 'newton-cg', 'newton-cholesky']
scores=[]
for s in solvers:
    lr=LogisticRegression(solver=s)
    pipe=Pipeline([('OHE',ct),('LR',lr)])
    pipe.fit(X_train,y_train)
    y_pred=pipe.predict(X_test)
    scores.append([s,accuracy_score(y_test,y_pred)])
df_score=pd.DataFrame(scores,columns=['Solver','Accuracy'])
df_score.sort_values('Accuracy',ascending=False)
```

```python
df=pd.read_csv(r"E:\DataAnalytics\Cases\Image_Segmentation\Image_Segmentation.csv")
df.sample(5)
```

```python
X,y=df.drop('Class',axis=1),df['Class']
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.3,random_state=26,stratify=y)
```

```python
import numpy as np
solvers = ['lbfgs', 'saga', 'sag', 'newton-cg', 'newton-cholesky']
cs=np.linspace(0.001,10,20)
for s in solvers:
    for c in cs:
        lr=LogisticRegression(solver=s)
        lr.fit(X_train,y_train)
        y_pred=lr.predict(X_test)
        scores.append([s,c,accuracy_score(y_test,y_pred)])
df_score=pd.DataFrame(scores,columns=['Solver','C','Accuracy'])
```

```python
df_score.sort_values('Accuracy',ascending=False)
```
