```python
df=pd.read_csv(r"E:\DataAnalytics\Cases\human-resources-analytics\HR_comma_sep.csv")
df.head()
```

```python
X,y=df.drop('left',axis=1),df['left']
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.3,random_state=26,stratify=y)
```

```python
ohe=OneHotEncoder(sparse_output=False,drop='first').set_output(transform='pandas')
scaler=StandardScaler()
ct=make_column_transformer(
    (ohe,make_column_selector(dtype_include='object')),
    ('passthrough',make_column_selector(dtype_include='number')),
    verbose_feature_names_out=False
)
svm=SVC(kernel="linear")
pipe=Pipeline([('OHE',ct),('SCL',scaler),('SVC',svm)])
pipe.fit(X_train,y_train)
```

```python
y_pred=pipe.predict(X_test)
accuracy_score(y_test,y_pred)
```

```python
import numpy as np
cs=np.linspace(0.001, 10,20)
scores=[]
for c in tqdm(cs):
    svm=SVC(kernel="linear",C=c)
    pipe=Pipeline([('OHE',ct),('SCL',scaler),('SVC',svm)])
    pipe.fit(X_train,y_train)
    y_pred=pipe.predict(X_test)
    scores.append([c,accuracy_score(y_test,y_pred)])
df_score=pd.DataFrame(scores,columns=['C_values','Accuracy'])
df_score.sort_values('Accuracy',ascending=False)
```

```python
cs=np.linspace(0.001, 10,20)
scores=[]
for c in tqdm(cs):
    for d in range(2,5):
        svm=SVC(kernel="poly",C=c,degree=d)
        pipe=Pipeline([('OHE',ct),('SCL',scaler),('SVC',svm)])
        pipe.fit(X_train,y_train)
        y_pred=pipe.predict(X_test)
        scores.append([c,d,accuracy_score(y_test,y_pred)])
df_score=pd.DataFrame(scores,columns=['C_values','Degree','Accuracy'])
df_score.sort_values('Accuracy',ascending=False).head()
```
