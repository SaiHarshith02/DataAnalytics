# imputing the central tendencies manually

```python
# mean
mu=dict(job.mean().round(2))
job.fillna(mu)
```

```python
# median
mu=dict(job.median().round(2))
job.fillna(mu)
```

# Imputing Using Transformer or SimpleImputer 

```python
from sklearn.impute import SimpleImputer
imputer=SimpleImputer(strategy='mean').set_output(transform='pandas')
imputer.fit(job) # just calculates the stats  and will not fill into null places
imputer.statistics_
imputer.transform(job).round(2) # transforms the data set by replacing the nulls


```
