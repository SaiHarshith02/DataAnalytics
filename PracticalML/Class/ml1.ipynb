```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
pizza=pd.read_csv(r'E:\DataAnalytics\Datasets\pizza.csv')
pizza.head()
```

```python
pizza.corr()
```

```python
plt.scatter(x=pizza['Promote'],y=pizza['Sales'])
plt.xlabel('Promotional Expences')
plt.ylabel('Product Sales')
plt.title('Promotion Expences VS Product Sales')
plt.show();
```

```python
X=pizza[['Promote']]
y=pizza['Sales']
```

```python
from param import random_seed
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(X,y,train_size=0.8,random_state=1)
model=LinearRegression()
model.fit(X_train,y_train)

```

```python
print('Coefficient:',model.coef_)
print('Intercept: ',model.intercept_)
```

```python
test_res=model.predict(X_test)
```

```python
plt.plot(test_res,y_test)
```

```python
df=pd.read_csv(r'../../Datasets\boston.csv')
df.head()
```

```python
sns.heatmap(df.corr(),annot=True,fmt='.2f',annot_kws={'size':6})
```

```python
from sklearn.metrics import root_mean_squared_error
X,y=df.drop('medv',axis=1),df['medv']
X_train,X_test,y_train,y_test=train_test_split(X,y,random_state=26,test_size=0.3)
model=LinearRegression()
model.fit(X_train,y_train)
y_pred=model.predict(X_test)
root_mean_squared_error(y_test,y_pred)
```
