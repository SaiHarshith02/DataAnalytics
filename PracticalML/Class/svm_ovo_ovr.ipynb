```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.compose import make_column_transformer,make_column_selector
from tqdm import tqdm
```

```python
df=pd.read_csv(r"E:\DataAnalytics\Cases\Vehicle Silhouettes\Vehicle.csv")
df.head()
```

```python
X,y=df.drop('Class',axis=1),df.Class
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.3,random_state=26,stratify=y)
```

```python
import numpy as np
cs=np.linspace(0.001, 10,20)
scaler=StandardScaler()
scores=[]
dfs=['ovo','ovr']
for c in tqdm(cs):
    for f in dfs:
        svm=SVC(kernel="linear",C=c,decision_function_shape=f)
        pipe=Pipeline([('SCL',scaler),('SVC',svm)])
        pipe.fit(X_train,y_train)
        y_pred=pipe.predict(X_test)
        scores.append([c,f,accuracy_score(y_test,y_pred)])
df_score=pd.DataFrame(scores,columns=['C_values','Decision','Accuracy'])
df_score.sort_values('Accuracy',ascending=False)
```

## Radial Bias Function as Kernel

```python
import numpy as np
cs=np.linspace(0.001, 10,20)
gs=np.linspace(0.1,10,20)
scaler=StandardScaler()
scores=[]
dfs=['ovo','ovr']
for c in tqdm(cs):
    for f in dfs:
        for g in gs:
            svm=SVC(kernel="rbf",C=c,gamma=g,decision_function_shape=f)
            pipe=Pipeline([('SCL',scaler),('SVC',svm)])
            pipe.fit(X_train,y_train)
            y_pred=pipe.predict(X_test)
            scores.append([c,f,g,accuracy_score(y_test,y_pred)])
df_score=pd.DataFrame(scores,columns=['C_values','Gamma_Values','Decision','Accuracy'])
df_score.sort_values('Accuracy',ascending=False)
```
