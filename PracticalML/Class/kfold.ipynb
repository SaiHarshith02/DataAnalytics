```python
import numpy as np
import pandas as pd 
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score,r2_score
from sklearn.tree import DecisionTreeClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import AdaBoostClassifier
from tqdm import tqdm
from sklearn.preprocessing import OneHotEncoder,LabelEncoder
from sklearn.compose import make_column_transformer, make_column_selector
from sklearn.model_selection import StratifiedKFold,cross_val_score
from sklearn.ensemble import GradientBoostingClassifier
import warnings
warnings.filterwarnings("ignore")
```

```python
df=pd.read_csv(r"E:\DataAnalytics\Cases\Vehicle Silhouettes\Vehicle.csv")
df.head()
```

```python
X,y=df.drop('Class',axis=1),df['Class']
le=LabelEncoder()
le_y=le.fit_transform(y)
kfold=StratifiedKFold(n_splits=5,shuffle=True,random_state=26)
gbm=GradientBoostingClassifier(random_state=26)
```

## Accuracy Scores of each Folds

```python
results=cross_val_score(gbm,X,le_y,cv=kfold) # default scoring='accuracy_score'
results
```

```python
import numpy as np
from tqdm import tqdm
from sklearn.ensemble import GradientBoostingRegressor
learning_rate=np.linspace(0.1,0.9,10)
max_depth=[None,2,3,4]
n_estimators=[25,50,100]
scores=[]

for l in tqdm(learning_rate):
    for m in max_depth:
        for n in n_estimators:
            gbc=GradientBoostingClassifier(random_state=26,learning_rate=l,max_depth=m,n_estimators=n)
            results=cross_val_score(gbm,X,le_y,cv=kfold,scoring='neg_log_loss') # default scoring='accuracy_score'
            scores.append([l,m,n,results.mean()])
df_scores=pd.DataFrame(scores,columns=['Learning_Rate','Max_Depth','N_Estimators','fold_best_log_loss'])
df_scores.sort_values(by='fold_best_log_loss',ascending= False)
```

```python
df_scores.sort_values(by='fold_best_log_loss',ascending= False)
```

```python
solvers = ['lbfgs', 'saga', 'sag', 'newton-cg', 'newton-cholesky']
from tqdm import tqdm
from sklearn.ensemble import GradientBoostingRegressor

scores=[]

for l in tqdm(solvers):
    lr=LogisticRegression(solver=l)
    results=cross_val_score(lr,X,le_y,cv=kfold,scoring='neg_log_loss') # default scoring='accuracy_score'
    scores.append([l,results.mean()])
df_scores=pd.DataFrame(scores,columns=['Solvers','log_loss'])
df_scores.sort_values(by='log_loss',ascending=False)
```

```python
solvers = ['lbfgs', 'saga', 'sag', 'newton-cg', 'newton-cholesky']
from tqdm import tqdm
from sklearn.ensemble import GradientBoostingRegressor
Cs=np.linspace(0.01,3,5)
scores=[]

for l in tqdm(solvers):
    for c in Cs:
        lr=LogisticRegression(solver=l,C=c)
        results=cross_val_score(lr,X,le_y,cv=kfold,scoring='neg_log_loss') # default scoring='accuracy_score'
        scores.append([l,c,results.mean()])
df_scores=pd.DataFrame(scores,columns=['Solvers','C_value','log_loss'])
df_scores.sort_values(by='log_loss',ascending=False)
```
