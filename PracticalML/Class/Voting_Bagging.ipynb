## -----Voting Ensembling Technique-----

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier 
from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score
from sklearn.ensemble import VotingClassifier
from tqdm import tqdm
import numpy as np
import warnings
warnings.filterwarnings("ignore")
```

```python
df=pd.read_csv(r"E:\DataAnalytics\Cases\Wisconsin\BreastCancer.csv",index_col=0)
df.head()
```

```python
X,y=df.drop('Class',axis=1),df['Class']
X_train,X_test,y_train,y_test=train_test_split(X,y,random_state=26,test_size=0.3,stratify=y)
```

```python
dtc=DecisionTreeClassifier(random_state=26)
LogReg=LogisticRegression()
knn=KNeighborsClassifier()
vc=VotingClassifier([('TREE',dtc),('LOGR',LogReg),('KNN',knn)])
```

```python
vc.fit(X_train,y_train)
```

```python
y_pred=vc.predict(X_test)
accuracy_score(y_test,y_pred)
```

```python
vc.named_estimators_
```

```python
vc.named_estimators_['TREE'].predict(X_test)
```

```python
scores=[]
for d in tqdm([None,2,3,4,5,6,7,8,9]):
    for s in ['lbfgs','sag','saga','liblinear']:
        for k in [2,3,4,5,10,13,15,17,20]:
            dtc=DecisionTreeClassifier(random_state=26,max_depth=d)
            LogReg=LogisticRegression(solver=s)
            knn=KNeighborsClassifier(n_neighbors=k)
            vc=VotingClassifier([('TREE',dtc),('LOGR',LogReg),('KNN',knn)],voting='soft') 
            #default voting tech is hard which returns mostfrequent output
            #soft voting technique will average the probabilies of each class 
            vc.fit(X_train,y_train)
            y_pred=vc.predict(X_test)
            scores.append([d,s,k,accuracy_score(y_test,y_pred)])
            
```

```python
df_scores=pd.DataFrame(scores,columns=['Depth','Solver','Neighbors','Accuracy'])
df_scores.sort_values('Accuracy',ascending=False)
```

```python
y_test_enc=np.where(y_test=='Malignant',1,0)
y_pred=vc.named_estimators_['TREE'].predict(X_test)
print('Tree Acc = ',accuracy_score(y_test_enc,y_pred))
y_pred=vc.named_estimators_['LOGR'].predict(X_test)
print('Logistic Acc = ',accuracy_score(y_test_enc,y_pred))
y_pred=vc.named_estimators_['KNN'].predict(X_test)
print('Knn Acc = ',accuracy_score(y_test_enc,y_pred))
```

# ------Booting Sampling Technique-------

```python
from sklearn.ensemble import BaggingClassifier
```

```python
df=pd.read_csv(r"E:\DataAnalytics\Cases\Wisconsin\BreastCancer.csv",index_col=0)
df.head()
X,y=df.drop('Class',axis=1),df['Class']
X_train,X_test,y_train,y_test=train_test_split(X,y,random_state=26,test_size=0.3,stratify=y)
```

```python
lr=LogisticRegression()
bagg=BaggingClassifier(estimator=lr,n_estimators=15,random_state=26)
bagg.fit(X_train,y_train)
```

```python
y_pred=bagg.predict(X_test)
accuracy_score(y_test,y_pred)
```

```python
est=[lr,knn,dtc]
n_samples=[10,25,50]
scores=[]
for e in est:
    for n in n_samples:
        bagg=BaggingClassifier(estimator=e,n_estimators=n,random_state=26)
        bagg.fit(X_train,y_train)
        y_pred=bagg.predict(X_test)
        scores.append([e,n,accuracy_score(y_test,y_pred)])
df_scores=pd.DataFrame(scores,columns=['Estimator','N','Accuracy'])
```

```python
df_scores.sort_values('Accuracy',ascending=False)
```

# Exercise -- on concreetDataset with ridge, lasso, Dtreeregressor

```python
from sklearn.ensemble import BaggingRegressor
```

```python
df=pd.read_csv(r"E:\DataAnalytics\Cases\Concrete_Strength\Concrete_Data.csv")
df.head()
```

```python
X,y=df.drop('Strength',axis=1),df['Strength']
X_train,X_test,y_train,y_test=train_test_split(X,y,random_state=26,test_size=0.3)
```

```python
lr=LinearRegression()
rid=Ridge()
lasso=Lasso()
tree=DecisionTreeRegressor(random_state=26)
```

```python
est=[lr,rid,lasso,tree]
n_samples=[10,25,50]
scores=[]
for e in est:
    for n in n_samples:
        bagg=BaggingRegressor(estimator=e,n_estimators=n,random_state=26)
        bagg.fit(X_train,y_train)
        y_pred=bagg.predict(X_test)
        scores.append([e,n,r2_score(y_test,y_pred)])
df_scores=pd.DataFrame(scores,columns=['Estimator','N','R2_Score'])
df_scores.sort_values('R2_Score',ascending=False)
```
