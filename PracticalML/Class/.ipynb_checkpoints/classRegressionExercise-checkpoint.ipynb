```python
import pandas as pd
from sklearn.linear_model import LinearRegression,ElasticNet,Ridge,Lasso
from sklearn.model_selection import train_test_split
from sklearn.metrics import r2_score,root_mean_squared_error
```

```python
df=pd.read_csv(r'../../Datasets\Exp_Salaries.csv')
df.head()
```

```python
X=df.drop('Salary',axis=1)
y=df['Salary']
```

```python
from xarray.util.generate_ops import inplace
from sklearn.preprocessing import OneHotEncoder
encoder=OneHotEncoder(sparse_output=False,drop='first').set_output(transform='pandas')
X_cat=encoder.fit_transform(X.select_dtypes('object'))
X=pd.concat([X.select_dtypes('number'),X_cat],axis=1)
X
```

```python
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.30,random_state=26)
```

```python
model=LinearRegression()
model.fit(X_train,y_train)
```

```python
y_pred=model.predict(X_test)
```

```python
print(r2_score(y_test,y_pred))
print(root_mean_squared_error(y_test,y_pred))
```

```python
# Ridge
from bokeh.layouts import column
import numpy as np
alpha=np.linspace(0.001,10,20)
scores=[]
for x in alpha:
    model=Ridge(alpha=x)
    model.fit(X_train,y_train)
    y_pred=model.predict(X_test)
    scores.append([x,r2_score(y_test,y_pred),root_mean_squared_error(y_test,y_pred)])
ridgeScr=pd.DataFrame(scores,columns=['Alpha','r2_Score','RMSE_Score']).sort_values('r2_Score',ascending=False)
ridgeScr.head()

```

```python
# lasso
alpha=np.linspace(0.001,10,20)
scores=[]
for x in alpha:
    model=Lasso(alpha=x)
    model.fit(X_train,y_train)
    y_pred=model.predict(X_test)
    scores.append([x,r2_score(y_test,y_pred),root_mean_squared_error(y_test,y_pred)])
lassoScr=pd.DataFrame(scores,columns=['Alpha','r2_Score','RMSE_Score']).sort_values('r2_Score',ascending=False)
lassoScr.head()
```

```python
# ElasticNet
alpha=np.linspace(0.001,10,20)
l1_ratio=np.linspace(0.001,1,20)
scores=[]
for x in alpha:
    for y in l1_ratio:
        model=ElasticNet(alpha=x,l1_ratio=y)
        model.fit(X_train,y_train)
        y_pred=model.predict(X_test)
        scores.append([x,r2_score(y_test,y_pred),root_mean_squared_error(y_test,y_pred)])
elastScr=pd.DataFrame(scores,columns=['Alpha','r2_Score','RMSE_Score']).sort_values('r2_Score',ascending=False)
elastScr.head()
```

```python
from sklearn.compose import make_column_selector,make_column_transformer
ohc=OneHotEncoder(sparse_output=False,drop='first').set_output(transform='pandas') 
ct= make_column_transformer((ohc, make_column_selector(dtype_include=object)),("passthrough", make_column_selector(dtype_include=['int64', 'float64'])), verbose_feature_names_out=False)
ct=ct.set_output(transform='pandas')
X_ohe= ct.fit_transform(X)
X_ohe
```
