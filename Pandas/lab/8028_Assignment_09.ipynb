Independent Two-Sample t-Test
Test difference in mean mpg between 4-cylinder and 6-cylinder cars
Perform t-test and display results
Conclusion on hypothesis

Expected output:
Hypotheses and sample means
t-test statistics and confidence interval
p-value significance conclusion
Interpretation statement

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
df=sns.load_dataset('mpg')
df.info()
```

```python
meanof_4cyl=df[df['cylinders']==4]['mpg']
meanof_6cyl=df[df['cylinders']==6]['mpg']
print('Avg miles per gallon of 4 cylinder car',meanof_4cyl.mean().round(3))
print('Avg miles per gallon of 6 cylinder car',meanof_6cyl.mean().round(3))
```

```python
from scipy.stats import ttest_ind
result=ttest_ind(meanof_4cyl,meanof_6cyl,alternative='two-sided')
print('T-Statistic :-',result.statistic)
print('P-Value:-',p_value:=result.pvalue)
```

```python
alpha=0.05
print('Null Hypothesis (H0)     :μ(4_cylinder)=μ(6_cylinder)')
print('Null Hypothesis (H1)     :μ(4_cylinder)!=μ(6_cylinder)')
```

```python
if p_value<alpha:
    print('Reject H0: Evidence that both mean are Diffrent')
else:
    print('Failed To Reject H0: Evidence that both mean are same')
```
