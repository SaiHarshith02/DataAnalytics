```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
```

```python
programms=[ 'Java', 'Python', 'PHP', 'JavaScript', 'C#', 'C++']
Popularity=[22.2, 17.6, 8.8, 8, 7.7, 6.7]
```

```python
plt.figure(figsize=(6,5))
plt.bar(programms,Popularity,color='blue',edgecolor='black',alpha=0.9)
ax = plt.gca()
ax.grid(which='major',color='red')
ax.minorticks_on()
ax.grid(which='minor',ls='--')
ax.set_axisbelow(True)
plt.xlabel('Languages')
plt.ylabel('Popularity')
plt.title('Popularity of Programming Language\n Worldwide, Oct 2017 Compared to a year ago')
plt.margins(x=0)
plt.show();
```

```python
plt.figure(figsize=(6,5))
plt.barh(programms,Popularity,color='Green',edgecolor='black',alpha=0.9)
ax = plt.gca()
ax.grid(which='major',color='red')
ax.minorticks_on()
ax.grid(which='minor',ls='--')
ax.set_axisbelow(True)
plt.xlabel('Popularity')
plt.ylabel('Languages')
plt.title('Popularity of Programming Language\n Worldwide, Oct 2017 Compared to a year ago')
plt.margins(y=0)
plt.show();
```

```python
plt.figure(figsize=(6,5))
plt.bar(programms,Popularity,edgecolor='black',alpha=0.6)
ax = plt.gca()
ax.grid(which='major',color='red')
ax.minorticks_on()
ax.grid(which='minor',ls='--')
ax.set_axisbelow(True)
plt.xlabel('Languages')
plt.ylabel('Popularity')
plt.title('Popularity of Programming Language\n Worldwide, Oct 2017 Compared to a year ago')
plt.margins(x=0)
ax.invert_yaxis()
plt.show();
```
