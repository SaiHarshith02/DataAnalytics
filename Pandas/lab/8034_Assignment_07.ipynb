1. Write a Python program to draw a line with suitable label in the x axis, y

2. Write a Python program to draw a line using given axis values with suitable

13. Write a Python program to display the grid and draw line charts of the
closing value of Alphabet Inc. between October 3, 2016 to October 7, 2016.

```python
dates = ['2016-10-03', '2016-10-04', '2016-10-05', '2016-10-06', '2016-10-07']
values = [772.559998, 776.429993, 776.469971, 776.859985, 775.080017]
plt.plot(dates,values,marker='o',color='r')
plt.grid(color='blue',alpha=0.3)
plt.yticks()
plt.margins(x=0,y=0)
plt.title('Closing stock value of Alphabet inc.')
plt.xlabel('Date')
plt.xlabel('Closing Value')
plt.plot();
```

14. Write a Python program to display the grid and draw line charts of the
closing value of Alphabet Inc. between October 3, 2016 to October 7, 2016.
Customized the grid lines with rendering with a larger grid (major grid) and a
smaller grid (minor grid).Turn on the grid but turn off ticks.

```python
dates = ['2016-10-03', '2016-10-04', '2016-10-05', '2016-10-06', '2016-10-07']
values = [772.559998, 776.429993, 776.469971, 776.859985, 775.080017]
plt.plot(dates,values,marker='o',color='r',markeredgecolor='b')
area=plt.gca()
area.minorticks_on()
area.grid(which='major',color='r')
area.grid(which='minor',ls='dotted')
plt.yticks()
plt.margins(x=0,y=0)
plt.title('Closing stock value of Alphabet inc.')
plt.xlabel('Date')
plt.xlabel('Closing Value')
plt.plot();
```

15. Write a Python program to create multiple plots.
The code snippet gives the output shown in the following screenshot:

```python
layout = [
    ['top', 'top', 'top'],
    ['bottom1', 'bottom2', 'bottom3']
]
fig, axs = plt.subplot_mosaic(layout, figsize=(8, 5))
for ax in axs.values():
    ax.set_xticks([])
    ax.set_yticks([])

plt.tight_layout()
plt.show()
```
