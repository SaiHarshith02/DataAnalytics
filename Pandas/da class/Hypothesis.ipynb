```python
import pandas as pd
from scipy.stats import zscore,norm
```

```python
mu=500 #clamed population mean
x_bar=503.4 # sample mean
s=10 #sample Std
n=25 #sample size 
alpha=0.05 # significence Level
```

```python
# Standerd error
se=s/n**0.5
#z test statistic
z=(x_bar-mu)/se
```

```python
# Critical value
# percent point function - probability -|> z score
critical_val=norm.ppf(1-alpha/2)
```

```python
p_value=2*norm.sf(abs(z))
print('population mean = ',mu)
print('Sample mean = ',x_bar)
print('Std = ',s)
print('Alpha =',alpha)
print('Standard error= ',se)
print('critical value= ',critical_val)
print('p-value = ',p_value)
```

```python
if z>critical_val:
    print('Critical value method rejected Ho')
else:
    print('Critical value method do not reject Ho')
```

```python
if p_value<alpha:
    print('p value method rejected Ho')
else:
    print('p value method do not reject Ho')
```
