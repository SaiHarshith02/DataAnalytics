```python
import numpy as np
```

```python
def roll_dice():
    # simulation of rolling dice twice,minimum number we can get is 1 and maximum 6
    # add the simulation results so possible values will be (1,1) or(1,2)........ sum will be between  2 and 12
    #run multiple times to verify 
    return np.sum(np.random.randint(1,7,2))
print(roll_dice())
print(roll_dice())
print(roll_dice())
```

```python
# part 2  someone aqpproaches us saying i will give you 5 dollars if you get 7 and take 1 for other number than 7
# how do we know what will happen?
def monte_carlo_simulation(runs=1000):
    results=np.zeros(2)
    for _ in range(runs):
        if roll_dice()==7:
            results[0]+=1
        else:
            results[1]+=1
    print('profit or loss for 1000 trials',results[0]*5-results[1])
    return results
print(monte_carlo_simulation())
print(monte_carlo_simulation())
print(monte_carlo_simulation())
```

```python
# part 3 lets do monte 1000 and in each trial 1000 runs like(1000) vals in each trial
results=np.zeros(1000)
for i in range(1000):
    results[i]=monte_carlo_simulation()[0]
print(results)
```

```python
# let us plot it
import matplotlib.pyplot as plt
fig,ax=plt.subplots()
ax.hist(results,bins=15)
plt.show()
```

```python
print(results.mean())
print(results.mean()*5)
print(results.mean()*4.75)
print(results.mean()*6)
```
