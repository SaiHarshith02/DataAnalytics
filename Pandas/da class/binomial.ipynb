```python
#poisson distribution
# 15 cars pass a toll plaza every hour. chance that exactly 13 pass next hour?
from scipy.stats import poisson
x=13
mu=15
print(poisson.pmf(x, mu))
# we read 10 pages an hour on average. chance of reading exactly 8 pages next hour?
x=8
mu =10
print(poisson.pmf(x, mu))
# 5 accidents a month at a dangerous corner 
mu=5
for x in range(5):
    print(poisson.pmf(x, mu))

print(poisson.cdf(2, mu))
print(1-poisson.cdf(3, mu))
```
