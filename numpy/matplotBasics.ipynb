```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv(r"C:\Users\PGCP-AI\Downloads\tips.csv")
df
```

```python
plt.figure(figsize=(500))
plt.hist(df['total_bill'],bins=10,color='skyblue',edgecolor='black')
plt.title('Distribution of total bill')
plt.xlabel('TOTAL BILL')
plt.ylabel('FREQUENCY')
plt.show()
```

```python
avgtip_gender=df.groupby('gender')['tip'].mean()
plt.figure(figsize=(6,4))
plt.bar(avgtip_gender.index,avgtip_gender.values,color=['pink','lightblue'])
plt.title('Average tip by gender')
plt.ylabel('Average Tip')
plt.show
```

```python
plt.figure(figsize=(6,4))
plt.scatter(df['total_bill'], df['tip'], color='orange', edgecolor='black')
plt.title('Total Bill vs Tip')
plt.xlabel('Total Bill')
plt.ylabel('Tip')
plt.show()
```

```python
plt.figure(figsize=(15,5))
plt.plot(df.index,df['total_bill'], marker='o', linestyle='-', color='purple')
plt.title('Total Bill across records')
plt.xlabel('Total Bill')
plt.ylabel('Tip')
plt.show()
```
