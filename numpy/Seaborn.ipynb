```python
import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv(r'C:\Users\PGCP-AI\Downloads\tips.csv')
```

```python
plt.figure(figsize=(6,4))
sns.histplot(df['total_bill'], bins=10, color='darkgreen')
plt.title('Distribution of Total Bill')
plt.show()
```

```python
plt.figure(figsize=(6,4))
sns.boxplot(x='gender', y='tip', data=df, hue='gender')
plt.title('Tip by Gender')
plt.show()
```

```python
plt.figure(figsize=(6,4))
sns.countplot(x='smoker', data=df, hue='smoker')
plt.title('Count of Smokers vs Non- Smokers')
plt.show()
```

```python
plt.figure(figsize=(6,4))
sns.regplot(x='total_bill', y='tip', data=df, color='red')
plt.title('Total Bill vs Tip with Regression Line')
plt.show()
```

```python
plt.figure(figsize=(6,4))
sns.barplot(x='day', y='tip', data=df)
plt.title('Average Tip by day')
plt.show()
```

```python
plt.figure(figsize=(12,5))
avg_tip_day=df.groupby('day')['tip'].mean()
plt.subplot(1,2,1)
plt.bar(avg_tip_day.index,avg_tip_day.values,color=['lightblue','lightgreen','pink','orange'])
plt.title('Avg Tip by Day')
plt.ylabel('Average Tips')
# plot 2
smoker_counts=df['smoker'].value_counts()
plt.subplot(1,2,2)
plt.pie(smoker_counts,labels=smoker_counts.index,autopct='%2.2f%%',colors=['gold','blue'])
plt.title('Smoker vs Non smokers')
plt.tight_layout()
plt.show()
```

```python
fig,axes=plt.subplots(1,2,figsize=(12,5))
axes[0].plot(df['total_bill'].iloc[:20],marker='o',ls='-',color='purple')
axes[0].set_title('Total Bill (first 20 records)')
axes[0].set_xlabel('Record Index')
axes[0].set_ylabel('Total Bill')
# plot 2
avg_tip_gender=df.groupby('gender')['tip'].mean()
axes[1].barh(avg_tip_gender.index,avg_tip_gender.values,color=['pink','lightblue'])
axes[1].set_title('Avg Tip By Gender')
axes[1].set_xlabel('Average Tip')
plt.tight_layout()
plt.show()
```

```python
nums=[1,2,4,1,2,5,1,2]
c1=None
c2=None
res=[]
count1,count2=0,0
for ele in nums:

    if ele ==c1:
        count1+=1

    elif ele==c2:
        count2+=1

    else:
        if count1==0:
            c1=ele
            count1=1

        elif count2==0:
            c2=ele
            count2=1

        else:
            count2-=1
            count1-=1

v1,v2=0,0

for ele in nums:
    if ele ==c1:
        v1+=1

    elif ele==c2:
        v2+=1

if v1>len(nums)//3:
    res.append(c1)

if v2>len(nums)//3 and c2 !=c1:
    res.append(c2)

print(res)
```
