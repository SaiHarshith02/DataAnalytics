```python
import seaborn as sns
airline_df=pd.read_csv(r"C:\Users\PGCP-AI\Downloads\airline_passenger_satisfaction.csv")
# print(airline_df.head())
on_board_service=airline_df['On-board Service']
mean_rating=on_board_service.mean()
median_rating=on_board_service.median()
skewness_rating=on_board_service.skew()
kurtosis_rating=on_board_service.kurt()
print(on_board_service.value_counts().sort_index())
print(f'mean onboard service:{mean_rating:.2f}')
print(f'median onboard service:{median_rating:.2f}')
print(f'skewness onboard service:{skewness_rating:.2f}')
print(f'kurtosis onboard service:{kurtosis_rating:.2f}')
```

```python
plt.figure(figsize=(10, 4))
sns.countplot(x='On-board Service', data=airline_df, palette='Set2')
plt.axvline(x=mean_rating, color='r', linestyle='--', label='Mean')
plt.axvline(x=median_rating, color='g', linestyle='--', label='Median')
plt.title('Distribution of On-board Service\n skewness: {:.2f}, kurtosis: {:.2f}'.format(skewness_rating, kurtosis_rating))
plt.xlabel('On-board Service Rating')
plt.ylabel('Passenger Count')
plt.legend()
plt.tight_layout()
plt.show()
```
