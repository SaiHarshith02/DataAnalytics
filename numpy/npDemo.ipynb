```python
marks = np.array([69,26,78,91,54,34])
passed = marks >= 35
print(passed)
print(marks[passed])
print(marks[marks>70])
```

```python
print(marks[(marks>=50) & (marks<80)])
print(marks[(marks<355) | (marks>90)])
print(marks[~passed])
print(np.sum(marks<35))
```
