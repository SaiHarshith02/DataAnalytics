```python
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
from sklearn.discriminant_analysis import LinearDiscriminantAnalysis,QuadraticDiscriminantAnalysis 
import pandas as pd
from sklearn.pipeline import Pipeline
```

```python
df=pd.read_csv(r"E:\DataAnalytics\Cases\Vehicle Silhouettes\Vehicle.csv")
df.head()
```

```python
X,y=df.drop('Class',axis=1),df['Class']
```

```python
X_train,X_test,y_train,y_test=train_test_split(X,y,random_state=26,stratify=y,test_size=0.3)
```

```python
lca=LinearDiscriminantAnalysis()
```

```python
pipe=Pipeline([('LCA',lca)])
pipe.fit(X_train,y_train)
```

```python
y_pred=pipe.predict(X_test)
```

```python
print('Accuracy Score ->', accuracy_score(y_test,y_pred))
```

```python
qca=QuadraticDiscriminantAnalysis()
```

```python
pipe=Pipeline([('QCA',qca)])
pipe.fit(X_train,y_train)
```

```python
y_pred=pipe.predict(X_test)
```

```python
print('Accuracy Score ->', accuracy_score(y_test,y_pred))
```
